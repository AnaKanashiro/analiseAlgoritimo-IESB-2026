```python
from tictactoe import tictactoe as ttt
import time

board = ttt.initial_state()
user = ttt.X #meu user

def print_board(board):
  print()
  for row in board:
    print([cell if cell is not None else ' ' for cell in row])

while not ttt.terminal(board):
  #desenhar o tabuleiro
    print_board(board)

    cur_player = ttt.player(board)

    if cur_player == user:
      print("\nSua vez (X)")
      try:
        i = int(input("Linha (0, 1, 2): "))
        j= int(input("Coluna (0, 1, 2): "))

        board = ttt.result(board, (i, j))
      except:
        print("Jogada Inválida! Tente novamente")

    else:
      print("\nIA pensando...")
      time.sleep(1)
      move = ttt.minimax(board)
      if move is None:
        break
      board = ttt.result(board, move)
      print(f"IA jogou em: {move}")

for row in board:
  print ([cell if cell else " " for cell in row])
venceu = ttt.winner(board)
if venceu:
  print(f"\nVENCEDOR: {venceu}")
else:
  print("\nEMPATE")
```
